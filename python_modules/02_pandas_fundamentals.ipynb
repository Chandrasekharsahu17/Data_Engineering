## ✦ Topic 2.1 — DataFrame & Series Basics

````python
# 📖 STUDY EXAMPLE
data = {
    "customer_id": [1, 2, 3, 4],
    "name": ["Raj", "Priya", "Amit", "Sneha"],
    "city": ["Mumbai", "Delhi", "Mumbai", "Pune"],
    "spend": [5000, 3000, 8000, 1500],
    "is_active":[True,False,True,False]
    }
df = pd.DataFrame(data)
print(df)
print(type(df))          # DataFrame
print(type(df['name']))  # Series — a single column
````

## ✦ Topic 2.4 — Selecting Rows (loc & iloc)

````python
# 📖 STUDY EXAMPLE
df.loc[0]           # row by label/index
df.loc[0:2]         # rows 0 to 2 (inclusive)
df.iloc[0]           # row by position
df.iloc[0:3]         # first 3 rows by position
````

## ✦ Topic 2.5 — Filtering Rows (WHERE equivalent)

````python
# 📖 STUDY EXAMPLE — same as SQL: WHERE city = 'Mumbai'
mumbai_customers = df[df['city'] == 'Mumbai']

# Multiple conditions — same as SQL: WHERE city='Mumbai' AND is_active=True
mumbai_active = df[(df['city'] == 'Mumbai') & (df['is_active'] == True)]

# OR condition — same as SQL: WHERE city='Mumbai' OR city='Delhi'
mumbai_or_delhi = df[(df['city'] == 'Mumbai') | (df['city'] == 'Delhi')]
````

````python
# ⌨️ Q1: Filter customers from 'Bangalore' only
# YOUR CODE:
Bangalore_customers = df[df['city'] == 'Bangalore']

Bangalore_customers
````

````python
# ⌨️ Q2: Filter customers who are active AND from Mumbai
# YOUR CODE:

mumbai_active = df[(df['city'] == 'Mumbai') & (df['is_active'] == True)]

mumbai_active
````

````python
# ⌨️ Q3: Filter customers from Mumbai, Delhi, OR Bangalore (3 conditions)
# YOUR CODE:


customersss=df[(df['city']=="Mumbai") | (df['city']=="Delhi") | (df['city']=="Bangalore")]
customersss
````
