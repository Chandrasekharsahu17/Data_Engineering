## ✦ Topic 6.3 — Error Handling (try/except)

```
In real pipelines, files can be missing, columns can be
misnamed, data can be malformed. Code needs to not crash.
```


````python
# 📖 STUDY EXAMPLE
def safe_load(filepath):
    try:
        df = pd.read_csv(filepath)
        return df
    except FileNotFoundError:
        print(f"File not found: {filepath}")
        return None
    except Exception as e:
        print(f"Error loading file: {e}")
        return None

result = safe_load('nonexistent_file.csv')
````

## ✦ Topic 6.4 — Putting It Together: A Mini Pipeline

```
This is the exact pattern you'll use in your
GitHub Bronze/Silver/Gold DE project.
```


````python
# 📖 STUDY EXAMPLE — Bronze → Silver pattern
def bronze_to_silver(filepath):
    # Bronze: raw load
    df = pd.read_csv(filepath)

    # Silver: cleaning
    df = df.dropna(subset=['name', 'city'])
    df['is_active'] = df['is_active'].fillna(False)
    df['registration_date'] = pd.to_datetime(df['registration_date'])
    df = df.drop_duplicates()

    return df

silver_df = bronze_to_silver('customers.csv')
print(f"Rows after cleaning: {silver_df.shape[0]}")
````
