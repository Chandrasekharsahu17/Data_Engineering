## ✦ Topic 3.6 — JOIN equivalent

```
SQL:    SELECT c.name, o.total
        FROM customers c
        JOIN orders o ON c.customer_id = o.customer_id
Pandas: merged = customers_df.merge(orders_df,
                                     on='customer_id',
                                     how='inner')
        merged[['name', 'total']]

'inner' → INNER JOIN
'left'  → LEFT JOIN
'right' → RIGHT JOIN
'outer' → FULL OUTER JOIN
```
