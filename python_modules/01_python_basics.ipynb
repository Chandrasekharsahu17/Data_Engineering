## ✦ Topic 1.4 — Loops & Conditionals

````python
# 📖 STUDY EXAMPLE
prices = [500, 1200, 300, 4500, 800]
for price in prices:
    if price > 1000:
        print(price, "- expensive")
    else:
        print(price, "- affordable")
````

````python
# ⌨️ Q1: Loop through this list of ages. Print "minor" if under 18, "adult" otherwise
ages = [15, 22, 17, 34, 8, 45]
# YOUR CODE:
for age in ages:
    if age < 18:
        print(age, "minor")
    else:
        print(age, "adult")

````

````python
# ⌨️ Q2: Loop through a list of order totals. Sum only the totals above 1000.
# Print the final sum.
totals = [500, 1500, 2000, 800, 3000, 200]
# YOUR CODE:
a=0
for i in totals:
    if i>1000:
        a+=i
print(a)

len(totals)
````

## ✦ Topic 1.5 — Functions

````python
# 📖 STUDY EXAMPLE
def calculate_discount(price, discount_pct):
    return price - (price * discount_pct / 100)

result = calculate_discount(1000, 20)
print(result)
````

````python
# ⌨️ Q1: Write a function that takes a list of prices and returns the average
# Test it with: [500, 1200, 300, 4500, 800]
# YOUR CODE:
def avgg(a):
    b=0
    for i in a:
        b+=i
    c=b/len(a)
    return c
lst=[500, 1200, 300, 4500, 800]

avgg(lst)
````

````python
# ⌨️ Q2: Write a function is_active_customer(days_since_last_order)
# Returns True if days_since_last_order <= 30, else False
# Test with 15 and 45
# YOUR CODE:

def is_active_customer(days_since_last_order):
    if days_since_last_order <=30:
        return True
    else :
        return False 
    

is_active_customer(15)

is_active_customer(45)
````
