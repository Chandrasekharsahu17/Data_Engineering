## ✦ Topic 5.3 — pivot_table()

```
Similar to Excel pivot tables / SQL CASE WHEN + GROUP BY combos.
Reshapes data so one column's values become new columns.
```


````python
# 📖 STUDY EXAMPLE
pivot = df.pivot_table(
    values='spend',
    index='city',
    columns='is_active',
    aggfunc='sum'
)
# Result: rows = cities, columns = True/False, values = sum of spend
````

## ✦ Topic 5.4 — Date/Time Handling

```

```
