## ✦ Topic 4.2 — Handling Nulls

```
SQL:    UPDATE customers SET city = 'Unknown' WHERE city IS NULL
Pandas: df['city'] = df['city'].fillna('Unknown')

Pandas: df = df.dropna(subset=['city'])
```


````python
# 📖 STUDY EXAMPLE
df['city'].fillna('Unknown', inplace=True)   # fill with a value
df.dropna(subset=['spend'], inplace=True)    # drop rows with null spend
df.dropna(how='all')                          # drop rows only if ALL cols null
````

## ✦ Topic 4.4 — String Operations

## ✦ Topic 4.5 — Duplicates

```
SQL:    SELECT DISTINCT city FROM customers
Pandas: df['city'].unique()

SQL:    SELECT * FROM customers (remove exact duplicate rows)
Pandas: df.drop_duplicates()
```


````python
# 📖 STUDY EXAMPLE
df.duplicated().sum()         # count of duplicate rows
df.drop_duplicates()          # remove exact duplicate rows
df['city'].unique()           # distinct values
df['city'].nunique()          # count of distinct values
````
