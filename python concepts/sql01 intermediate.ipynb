````python
# ⌨️ SETUP — Run this first to create the database
import sqlite3
import pandas as pd

conn = sqlite3.connect('sql_practice.db')
cur = conn.cursor()

# Create tables
cur.executescript("""
DROP TABLE IF EXISTS customers;
DROP TABLE IF EXISTS orders;
DROP TABLE IF EXISTS products;
DROP TABLE IF EXISTS order_items;
DROP TABLE IF EXISTS employees;
DROP TABLE IF EXISTS departments;

CREATE TABLE customers (
    customer_id INTEGER PRIMARY KEY,
    name TEXT, city TEXT, age INTEGER, joined_year INTEGER
);

CREATE TABLE orders (
    order_id INTEGER PRIMARY KEY,
    customer_id INTEGER, order_date TEXT, status TEXT, total REAL
);

CREATE TABLE products (
    product_id INTEGER PRIMARY KEY,
    name TEXT, category TEXT, price REAL, stock INTEGER
);

CREATE TABLE order_items (
    item_id INTEGER PRIMARY KEY,
    order_id INTEGER, product_id INTEGER, qty INTEGER, unit_price REAL
);

CREATE TABLE employees (
    emp_id INTEGER PRIMARY KEY,
    name TEXT, dept_id INTEGER, salary REAL, manager_id INTEGER, hire_year INTEGER
);

CREATE TABLE departments (
    dept_id INTEGER PRIMARY KEY,
    dept_name TEXT, location TEXT
);
""")

# Insert data
cur.executemany("INSERT INTO customers VALUES (?,?,?,?,?)", [
    (1,'Raj','Mumbai',28,2020),(2,'Priya','Delhi',34,2018),
    (3,'Amit','Pune',22,2022),(4,'Sneha','Mumbai',29,2019),
    (5,'Dev','Bangalore',31,2021),(6,'Aisha','Delhi',26,2023),
    (7,'Kabir','Chennai',45,2017),(8,'Meera','Mumbai',38,2016)
])

cur.executemany("INSERT INTO orders VALUES (?,?,?,?,?)", [
    (1,1,'2024-01-10','completed',2500),(2,1,'2024-02-15','completed',800),
    (3,2,'2024-01-20','completed',5000),(4,3,'2024-03-01','pending',1200),
    (5,4,'2024-02-28','completed',3400),(6,4,'2024-03-10','cancelled',900),
    (7,5,'2024-01-05','completed',7500),(8,5,'2024-03-20','pending',2100),
    (9,7,'2024-02-10','completed',4200),(10,7,'2024-03-15','completed',1800),
    (11,8,'2024-01-25','completed',6300),(12,None,'2024-03-05','pending',500)
])

cur.executemany("INSERT INTO products VALUES (?,?,?,?,?)", [
    (1,'Laptop','Electronics',75000,15),(2,'Phone','Electronics',25000,42),
    (3,'Tablet','Electronics',40000,8),(4,'Earbuds','Accessories',3000,100),
    (5,'Keyboard','Accessories',2500,55),(6,'Monitor','Electronics',18000,20),
    (7,'Desk','Furniture',12000,10),(8,'Chair','Furniture',8000,25)
])

cur.executemany("INSERT INTO order_items VALUES (?,?,?,?,?)", [
    (1,1,2,1,25000),(2,1,4,2,3000),(3,2,5,1,2500),
    (4,3,1,1,75000),(5,4,3,1,40000),(6,5,2,2,25000),
    (7,6,4,3,3000),(8,7,1,1,75000),(9,8,6,1,18000),
    (10,9,2,1,25000),(11,10,5,2,2500),(12,11,1,1,75000)
])

cur.executemany("INSERT INTO departments VALUES (?,?,?)", [
    (1,'IT','Mumbai'),(2,'HR','Delhi'),(3,'Finance','Pune'),(4,'Marketing','Bangalore')
])

cur.executemany("INSERT INTO employees VALUES (?,?,?,?,?,?)", [
    (1,'Raj',1,85000,None,2019),(2,'Priya',1,72000,1,2020),
    (3,'Amit',1,65000,1,2021),(4,'Sneha',2,58000,None,2018),
    (5,'Dev',2,52000,4,2020),(6,'Aisha',3,90000,None,2017),
    (7,'Kabir',3,78000,6,2019),(8,'Meera',4,61000,None,2021),
    (9,'Rohan',4,55000,8,2022),(10,'Nisha',1,69000,1,2020)
])

conn.commit()
print("Database ready! Tables: customers, orders, products, order_items, employees, departments")
````
