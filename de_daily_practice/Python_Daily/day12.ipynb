```python
# ─── CONCEPT EXAMPLE ───────────────────────────────

import numpy as np, time

n = 1_000_000
lst = list(range(n))
arr = np.arange(n)

# Python list - slow
start = time.time()
total = sum(x**2 for x in lst)
print(f'Python: {time.time()-start:.3f}s')

# NumPy - fast
start = time.time()
total = np.sum(arr**2)
print(f'NumPy: {time.time()-start:.3f}s')
# NumPy is 10-100x faster because
# data stored in contiguous memory blocks
```
