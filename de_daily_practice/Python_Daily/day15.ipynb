```python
# ─── CONCEPT EXAMPLE ───────────────────────────────

import pandas as pd, numpy as np

messy = pd.DataFrame({
    'emp_id':  [1,2,2,3,4],
    'name':    ['raj','PRIYA','PRIYA',None,'amit'],
    'age':     [28,250,250,32,-5],
    'salary':  [55000,72000,72000,None,68000],
})

# Find problems first
print(messy.isnull().sum())
print(f'Dupes: {messy.duplicated().sum()}')

# Fix step by step
clean = messy.drop_duplicates(subset=['emp_id'])
clean = clean.dropna(subset=['name'])
clean['salary'] = clean['salary'].fillna(clean['salary'].median())
clean['name'] = clean['name'].str.strip().str.title()
clean = clean[(clean['age']>=0) & (clean['age']<=100)]
```
