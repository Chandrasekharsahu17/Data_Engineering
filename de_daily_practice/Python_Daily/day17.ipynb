```python
# ─── CONCEPT EXAMPLE ───────────────────────────────

import pandas as pd

orders = pd.DataFrame({'order_id':[1,2,3,4],'customer_id':[101,102,103,999],'total':[5000,3000,8000,2000]})
customers = pd.DataFrame({'customer_id':[101,102,103],'name':['Raj','Priya','Amit']})
# 999 has no customer record!

# INNER: only matching
inner = pd.merge(orders, customers, on='customer_id', how='inner')

# LEFT: all orders, NaN where no customer
left = pd.merge(orders, customers, on='customer_id', how='left')

# Find orphans
orphans = left[left['name'].isnull()]
print(f'Orders with no customer: {len(orphans)}')
```
