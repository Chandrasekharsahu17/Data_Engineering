```python
# ─── CONCEPT EXAMPLE ───────────────────────────────

import pandas as pd

df = pd.DataFrame([
    {'emp_id':1,'name':'Raj','dept':'IT','salary':85000},
    {'emp_id':2,'name':'Priya','dept':'HR','salary':72000},
    {'emp_id':3,'name':'Amit','dept':'IT','salary':92000},
])

# loc = LABEL based
df.loc[0]                           # row with index label 0
df.loc[df['dept']=='IT']            # filter by condition
df.loc[:, ['name','salary']]        # specific columns

# iloc = POSITION based
df.iloc[0]      # first row
df.iloc[0:2]    # rows 0 and 1

# Boolean filter
df[df['salary'] > 80000]
df[(df['dept']=='IT') & (df['salary']>80000)]
```
