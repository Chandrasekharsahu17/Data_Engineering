```python
# ─── CONCEPT EXAMPLE ───────────────────────────────

class DataPipeline:
    def __init__(self, name):
        self.name = name
        self.rows_processed = 0

    def extract(self, source):
        print(f'[{self.name}] Extracting from {source}')
        return []

    def transform(self, data):
        print(f'[{self.name}] Transforming {len(data)} rows')
        return data

    def load(self, data, dest):
        print(f'[{self.name}] Loading {len(data)} rows')
        self.rows_processed += len(data)

    def run(self, source, dest):
        data = self.extract(source)
        data = self.transform(data)
        self.load(data, dest)
        print('Pipeline complete')

p = DataPipeline('SalesPipeline')
p.run('sales.csv', 'company.db')
```
