```python
# ─── CONCEPT EXAMPLE ───────────────────────────────

import pandas as pd, sqlite3

def incremental_load(new_df, db_path, table):
    conn = sqlite3.connect(db_path)
    try:
        existing = pd.read_sql(f'SELECT order_id FROM {table}', conn)
        existing_ids = set(existing['order_id'])
        new_only = new_df[~new_df['order_id'].isin(existing_ids)]
    except:
        new_only = new_df  # table doesn't exist yet
    new_only.to_sql(table, conn, if_exists='append', index=False)
    print(f'Loaded {len(new_only)} new records')
    conn.close()
```
