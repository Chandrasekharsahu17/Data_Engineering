```python
# ─── CONCEPT EXAMPLE ───────────────────────────────

# Lambda = short one-line function
# lambda arguments: expression

add = lambda x, y: x + y
add(3, 4)   # 7

gst = lambda price: price * 1.18
gst(100)    # 118.0

# sorted() with lambda key
employees = [{'name':'B','dept_id':2,'salary':60000},
             {'name':'A','dept_id':1,'salary':80000},
             {'name':'C','dept_id':2,'salary':50000}]

# Sort by dept_id asc, then salary desc within dept
sorted_emps = sorted(employees,
    key=lambda e: (e['dept_id'], -e['salary']))
```
