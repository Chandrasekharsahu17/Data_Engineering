```python
# ─── CONCEPT EXAMPLE ───────────────────────────────

import pandas as pd, time

df = pd.DataFrame({'salary': range(1, 1000001)})

# SLOWEST: iterrows - never use on large data
start = time.time()
results = [row['salary']*1.18 for _,row in df.iterrows()]
print(f'iterrows: {time.time()-start:.3f}s')

# MEDIUM: apply
start = time.time()
results = df['salary'].apply(lambda x: x*1.18)
print(f'apply: {time.time()-start:.3f}s')

# FASTEST: vectorized
start = time.time()
results = df['salary'] * 1.18
print(f'vectorized: {time.time()-start:.3f}s')
```
