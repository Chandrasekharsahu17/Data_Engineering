```python
# ─── CONCEPT EXAMPLE ───────────────────────────────

import pandas as pd, sqlite3

def extract(fp):
    try:
        df = pd.read_csv(fp)
        print(f'Extracted {len(df)} rows')
        return df
    except FileNotFoundError:
        print('File not found')
        return pd.DataFrame()

def transform(df):
    before = len(df)
    df = df.drop_duplicates(subset=['emp_id'])
    df = df.dropna(subset=['emp_id'])
    df['annual'] = df['salary'] * 12
    print(f'Transform: {before} to {len(df)} rows')
    return df

def load(df, db, table):
    conn = sqlite3.connect(db)
    df.to_sql(table, conn, if_exists='replace', index=False)
    conn.close()
    print(f'Loaded {len(df)} rows')

def run(source, db, table):
    df = extract(source)
    if df.empty: return
    df = transform(df)
    load(df, db, table)
```
