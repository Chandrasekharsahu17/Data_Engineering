```python
# ─── CONCEPT EXAMPLE ───────────────────────────────

import numpy as np

a = np.array([1,2,3,4,5])

# Vectorized - no loop
a * 2          # [2,4,6,8,10]
a[a > 3]       # [4,5]
a[a < 0] = 0   # replace negatives

# 2D array
grid = np.array([[10,20,30],[40,50,60]])
grid.shape     # (2,3)

# axis: 0=across rows, 1=across columns
np.sum(grid, axis=0)   # col totals: [50,70,90]
np.sum(grid, axis=1)   # row totals: [60,150]
np.mean(grid, axis=1)  # row means: [20.0, 50.0]
```
