```python
# ─── CONCEPT EXAMPLE ───────────────────────────────

class TaskOrchestrator:
    def __init__(self):
        self.tasks = []
        self.results = {}

    def add_task(self, name, func, depends_on=None):
        self.tasks.append({'name':name,'func':func,'depends_on':depends_on})

    def run(self):
        for task in self.tasks:
            dep = task['depends_on']
            if dep and not self.results.get(dep,{}).get('success'):
                print(f'Skipping {task["name"]} (dep failed)')
                continue
            try:
                result = task['func']()
                self.results[task['name']] = {'success':True}
                print(f'PASS: {task["name"]}')
            except Exception as e:
                self.results[task['name']] = {'success':False}
                print(f'FAIL: {task["name"]}: {e}')
```
