```python
# ─── CONCEPT EXAMPLE ───────────────────────────────

class DataQualityFramework:
    def __init__(self):
        self.rules = []

    def add_rule(self, name, fn, severity='WARNING'):
        self.rules.append({'name':name,'fn':fn,'severity':severity})

    def run(self, df):
        results = []
        for rule in self.rules:
            passed = rule['fn'](df)
            results.append({'rule':rule['name'],'passed':passed,'severity':rule['severity']})
        return results

dqf = DataQualityFramework()
dqf.add_rule('no_null_ids', lambda df: df['emp_id'].notnull().all(), 'CRITICAL')
dqf.add_rule('valid_age',   lambda df: df['age'].between(0,100).all(), 'WARNING')
```
