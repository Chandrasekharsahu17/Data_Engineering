```python
# ─── CONCEPT EXAMPLE ───────────────────────────────

import pandas as pd

df = pd.DataFrame([
    {'emp_id':1,'dept':'IT','salary':85000},
    {'emp_id':2,'dept':'HR','salary':72000},
    {'emp_id':3,'dept':'IT','salary':92000},
    {'emp_id':4,'dept':'HR','salary':68000},
])

# Single agg
df.groupby('dept')['salary'].mean()

# Multiple aggs in one call
result = df.groupby('dept').agg(
    headcount  = ('emp_id','count'),
    avg_salary = ('salary','mean'),
    max_salary = ('salary','max'),
    total_bill = ('salary','sum'),
).reset_index()

# transform: keeps all rows
df['dept_avg'] = df.groupby('dept')['salary'].transform('mean')
df['above_avg'] = df['salary'] > df['dept_avg']
```
