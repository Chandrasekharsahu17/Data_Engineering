```python
# ─── CONCEPT EXAMPLE ───────────────────────────────

import pandas as pd

df = pd.DataFrame([
    {'name':'raj sharma','salary':85000,'hire_year':2019},
    {'name':'PRIYA MEHTA','salary':72000,'hire_year':2021},
])

# apply() on column
def salary_band(salary):
    if salary > 80000: return 'A'
    if salary > 60000: return 'B'
    return 'C'

df['band'] = df['salary'].apply(salary_band)

# apply(axis=1) = row-wise
df['exp'] = df.apply(lambda row: 2026-row['hire_year'], axis=1)

# String ops
df['name'] = df['name'].str.strip().str.title()
```
