```python
# ─── CONCEPT EXAMPLE ───────────────────────────────

def validate_df(df, schema):
    errors = []
    for col, rules in schema.items():
        if rules.get('required') and df[col].isnull().any():
            errors.append(f'{col}: has nulls')
        if 'min' in rules:
            bad = len(df[df[col] < rules['min']])
            if bad: errors.append(f'{col}: {bad} below minimum')
        if 'max' in rules:
            bad = len(df[df[col] > rules['max']])
            if bad: errors.append(f'{col}: {bad} above maximum')
        if 'allowed' in rules:
            bad = ~df[col].isin(rules['allowed'])
            if bad.any(): errors.append(f'{col}: invalid values')
    return {'valid': len(errors)==0, 'errors': errors}
```
