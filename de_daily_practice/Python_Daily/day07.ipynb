```python
# ─── CONCEPT EXAMPLE ───────────────────────────────

# try=attempt, except=handle error, else=if no error, finally=always

try:
    result = 10 / 0
except ZeroDivisionError:
    print('Cannot divide by zero')
except (TypeError, ValueError) as e:
    print(f'Error: {e}')
else:
    print('Success!')   # only if no error
finally:
    print('Always runs')

# Safe type conversion - common in DE
def safe_int(value, default=0):
    try:
        return int(value)
    except (TypeError, ValueError):
        return default

safe_int('123')  # 123
safe_int('abc')  # 0
safe_int(None)   # 0
```
