```python
# ─── CONCEPT EXAMPLE ───────────────────────────────

from functools import reduce

nums = [1, 2, 3, 4, 5]

# map: apply to EVERY element
doubled = list(map(lambda x: x*2, nums))    # [2,4,6,8,10]

# filter: KEEP where True
big = list(filter(lambda x: x > 3, nums))   # [4,5]

# reduce: combine into ONE result
# acc = accumulator, x = current item
total = reduce(lambda acc, x: acc + x, nums, 0)  # 15
# steps: 0+1=1, 1+2=3, 3+3=6, 6+4=10, 10+5=15

# DE pattern: total revenue
orders = [{'order_id':101,'revenue':5000},
          {'order_id':102,'revenue':3000}]
total = reduce(lambda acc, o: acc + o['revenue'], orders, 0)
```

## Q2

> One chain using map+filter+reduce:
orders=[{'id':1,'amount':5000,'valid':True},{'id':2,'amount':3000,'valid':False},{'id':3,'amount':7000,'valid':True},{'id':4,'amount':2000,'valid':True}]
Step1 filter: keep valid=True
Step2 map: extract amount
Step3 reduce: sum amounts


```python
# YOUR CODE:
orders=[{'id':1,'amount':5000,'valid':True},{'id':2,'amount':3000,'valid':False},{'id':3,'amount':7000,'valid':True},{'id':4,'amount':2000,'valid':True}]

z=reduce(lambda acc,x:acc+x,list(map(lambda x:x['amount'],filter(lambda a:a['valid'], orders))),0)
print(z)
```
