```python
# ─── CONCEPT EXAMPLE ───────────────────────────────

from collections import defaultdict

# Safe nested access
record = {'customer':{'name':'Raj','address':{'city':'Mumbai'}}}
city = record.get('customer',{}).get('address',{}).get('city','Unknown')
# 'Mumbai' - no KeyError even if key missing

# defaultdict: never raises KeyError
by_city = defaultdict(lambda: defaultdict(int))
orders = [{'prod':'Laptop','city':'Mumbai','rev':75000},
          {'prod':'Phone','city':'Delhi','rev':25000},
          {'prod':'Laptop','city':'Delhi','rev':75000}]

for o in orders:
    by_city[o['city']][o['prod']] += o['rev']
# {'Mumbai':{'Laptop':75000}, 'Delhi':{'Phone':25000,'Laptop':75000}}
```
