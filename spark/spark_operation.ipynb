```python
from pyspark.sql.types import StructType, StructField, IntegerType, StringType

data = [(1, "Alice", 25), (2, "Bob", 30), (3, "Charlie", 35)]

schema = StructType([
    StructField("id", IntegerType(), False),
    StructField("name", StringType(), False),
    StructField("age", IntegerType(), False)
])

df = spark.createDataFrame(data, schema)

df.show()
```

## Handellling data types in spark

```python
from pyspark.sql import SparkSession

spark=SparkSession.builder .appName('Handelling_datatypes_in_spark').getOrCreate()
```

```python
# Sample data
data = [
(1, "John Doe", "Bangalore", "2023-01-15", "152.75", "True"),
(2, "Jane Smith", "Delhi", "2023-05-20", "89.50", "False"),
(3, "Robert Brown", "Mumbai", "InvalidDate", "200.00", "True"),
(4, "Linda White", "Kolkata", "2023-02-29", None, "yes"), # Feb 29 invalid in 2023
(5, "Mike Green", "Chennai", "2023-08-10", "NaN", "1"), # NaN needs handling
(6, "Sarah Blue", "Hyderabad", "InvalidDate", "300.40", "No")
]
# Define column names
columns = ["id", "name", "city", "date", "amount", "is_active"]
# Create DataFrame
df = spark. createDataFrame(data, schema=columns)
# Show the DataFrame
df.show()
```

```python
df.printSchema()
```

```python
df.withColumn("id_double",df.id*2).show()
```

```python
from pyspark.sql.types import *
from pyspark.sql.functions import *
```

```python
df=df.withColumn('id', col('id').cast(IntegerType()))
```

```python
df.printSchema()
```

```python
df=df.withColumn("Upper_name",upper(df.name))
df.show()
```

```python
df.filter(df.city.startswith('B')).show()
```

```python
df=df.withColumn('amount', col('amount').cast(FloatType()))
df.show()
df.printSchema()
```

```python
df_fill=df.fillna({'amount':0})
df_fill.show()
```
