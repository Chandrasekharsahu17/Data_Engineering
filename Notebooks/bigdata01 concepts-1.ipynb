---

## ⌨️ BUILD 4 — Single Machine vs Distributed

> Fill in the comparison table:

| Factor | Single Machine | Distributed System |
|--------|---------------|--------------------|
| Storage Limit | limitied around 2 tb|can be more as per requirements more machine is more storage|
| Processing Speed |limited as it can handle very small limited computing power |based on requirement we can increase or decreae the processing speed|
| Fault Tolerance |very high once stopped your whole process stopped |here all are connected if even one fails others machines will work and work will be continued |
| Cost to Scale |very less |very high as many machine with same processeor storage will take |
| Example Tool |your own computer system | any bank server system  |

````python
# ⌨️ BUILD 4b — Simulate why a single machine fails
# Instructions:
# 1. Create a list of 10 million numbers (use range)
# 2. Time how long it takes to sum them on one "machine" (just this process)
# 3. Now split the list into 4 chunks (simulate 4 machines)
# 4. Sum each chunk separately, then combine
# 5. Print both times and explain which is faster and why

import time

# Step 1: Create a list of 10 million numbers
numbers = list(range(10_000_000))

# -----------------------------
# Single machine simulation
# -----------------------------
start = time.time()

single_sum = sum(numbers)

single_time = time.time() - start

# -----------------------------
# Simulate 4 machines
# -----------------------------
chunk_size = len(numbers) // 4

chunks = [
    numbers[0:chunk_size],
    numbers[chunk_size:2*chunk_size],
    numbers[2*chunk_size:3*chunk_size],
    numbers[3*chunk_size:]
]

start = time.time()

partial_sums = []

for chunk in chunks:
    partial_sums.append(sum(chunk))

distributed_sum = sum(partial_sums)

distributed_time = time.time() - start


# Results
print(f"Single machine sum: {single_sum}")
print(f"Single machine time: {single_time:.4f} sec")

print()

print(f"4-machine simulated sum: {distributed_sum}")
print(f"4-machine simulated time: {distributed_time:.4f} sec")

print()

# Explanation
if distributed_time < single_time:
    print("Chunked approach was faster.")
else:
    print("Single machine was faster.")


# becauae in single operarion it executed fast to make in chucnk it should have done in parallelnot splitting will work 
````
