````python
# ⌨️ BUILD 6b — Block Calculator
# Write a function that calculates:
# - Number of blocks needed
# - Size of the last block
# - Total storage WITH replication

import math

def calculate_hdfs_blocks(file_size_mb, block_size_mb=128, replication=3):
    num_blocks = math.ceil(file_size_mb / block_size_mb)
    last_block_size = file_size_mb % block_size_mb 
    total_storage = file_size_mb * replication
    print(num_blocks,last_block_size,total_storage)
    pass

# Test:
calculate_hdfs_blocks(300)      # 300MB file
calculate_hdfs_blocks(1024)     # 1GB file
calculate_hdfs_blocks(10240)    # 10GB file
````

````python
# ⌨️ BUILD 7b — Replication Tradeoff Calculator
# For replication factors 1 through 5, print:
# - Total storage needed
# - Estimated cost in USD
# - How many node failures can be tolerated

def replication_tradeoff(data_size_tb, cost_per_tb_usd=50):
    print(f"{'Replication':<15} {'Total Storage (TB)':<22} {'Cost (USD)':<20} {'Failures Tolerated'}")
    print("-" * 75)
    
    for replication in range(1,5):

        total_storage = data_size_tb * replication
        cost = total_storage * cost_per_tb_usd
        failures_tolerated = replication - 1
        print(f"{replication:<15} {total_storage:<22} {cost:<20} {failures_tolerated}")
    pass

replication_tradeoff(100)
````
