## ✦ Topic 5.2 — Subqueries in FROM (Derived Tables)

```sql
-- ⌨️ Q1: Get avg spend per customer, then find customers above that average
-- YOUR QUERY:
select customer_id,avg(total) from orders
group by customer_id
having avg(total) > (select avg(total) from orders)
```

```sql
-- ⌨️ Q2: Find top 3 products by total revenue using subquery in FROM
-- YOUR QUERY:
select name from products
where product_id in (select o2.product_id from orders o join order_items o2 
on o.order_id =o2.order_id
group by o2.product_id
order by sum(total) desc
limit 3
)
```

```sql
-- ⌨️ Q3: Find departments where highest paid employee earns > 2x the lowest paid
-- YOUR QUERY:
select dept_id,min(salary),max(salary) from employees 
group by dept_id
having 2*min(salary)<max(salary)
```

## ✦ Topic 5.3 — Correlated Subqueries

```sql
-- 📖 STUDY EXAMPLE — runs once per row in outer query
SELECT name, salary, dept_id
FROM employees e
WHERE salary > (
    SELECT AVG(salary) FROM employees WHERE dept_id = e.dept_id
);
```

```sql
-- ⌨️ Q1: Employees earning more than their own department average (use example above)
-- YOUR QUERY:
select name,salary from employees e
where salary > (
    SELECT AVG(salary) as x FROM employees WHERE dept_id = e.dept_id
);
```

```sql
-- ⌨️ Q2: Customers whose total spend is above average spend in their own city
-- YOUR QUERY:
select c.name ,sum(o.total)
from customers c 
join orders o 
on c.customer_id=o.customer_id
group by  c.customer_id, c.name, c.city
having sum(o.total) > (select avg(o.total)
                from customers c2 
            join orders o 
            on c2.customer_id=o.customer_id
            where c2.city=c.city)


```

## ✦ Topic 5.4 — IN, NOT IN, EXISTS, NOT EXISTS

```sql
-- ⌨️ Q1: Find customers who ordered at least one Electronics product (use IN)
-- YOUR QUERY:

select name from customers 
where customer_id in (select customer_id
                    from orders o 
                    join order_items o2 on 
                    o.order_id = o2.order_id
                    join products p on 
                    o2.product_id=p.product_id
                    where p.category = "Electronics"
                    )
```

```sql
-- ⌨️ Q2: Find customers who have NEVER placed an order (use NOT EXISTS)
-- YOUR QUERY:

select name from customers 
where customer_id NOT in  (select customer_id from orders )
```

## ✦ Topic 5.5 — Subquery vs JOIN vs CTE

```sql
select * from orders
```

```sql
-- ⌨️ Q1: Write the SAME query 3 ways: customers with more than 2 completed orders
--        Way 1: subquery  Way 2: JOIN  Way 3: CTE
-- YOUR QUERIES:
--        Way 1: subquery
select name from customers 
where customer_id in (select customer_id from orders 
where status = "completed"
group by customer_id
having  count(*)>2);


-- Way 2: JOIN 
select c.name from customers c
 join orders o on c.customer_id = o.customer_id
 where o.status = "completed"
 group by c.name
 having count (c.name) >2
 








```
