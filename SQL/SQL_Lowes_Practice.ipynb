### Q4 — WOS Calculation ⭐ Lowe's Core Metric
For each store + product in inventory:
- Calculate WOS = stock_units / weekly_sales

## 🪟 SECTION 5 — WINDOW FUNCTIONS (Q23-Q30)
### Most Important Section for This Interview
---

**Syntax reminder:**
```sql
FUNCTION() OVER (PARTITION BY col ORDER BY col)

ROW_NUMBER() → unique rank even on ties
RANK()       → same rank on ties, skips numbers
DENSE_RANK() → same rank on ties, no gaps
LAG(col)     → value from previous row
LEAD(col)    → value from next row
SUM() OVER (ORDER BY date) → running total
AVG() OVER (ROWS BETWEEN 1 PRECEDING AND CURRENT ROW) → moving avg
```

````python
%sql
-- YOUR CODE:
-- ROW_NUMBER() OVER (PARTITION BY s.product_id ORDER BY s.revenue DESC) AS rn

select p.name,s.city,s.store_id,sum(s2.revenue),
row_number() over (partition by p.name order by sum(s2.revenue)) as rank
from stores s
join sales s2 on s.store_id=s2.store_id
join products p on s2.product_id=p.product_id
group by p.name,s.city,s.store_id

````

### Q24 — TOP N per Group ⭐ Most Common Interview Question
Find the TOP 2 stores by revenue for each product.
Use ROW_NUMBER in a CTE, then filter WHERE rn <= 2.

````python
%sql
select
name,store_id,rn as rank
from 
(select 
p.name,s.store_id,
row_number() over (partition by p.name order by sum(s2.revenue) desc ) as rn
from stores s
join sales s2 on 
s.store_id=s2.store_id
join products p on s2.product_id=p.product_id
group by p.name,s.store_id) as x
where x.rn<=2
order by name

````

````python
%sql
-- ANSWER Q1 — INNER JOIN Tools
SELECT st.city, p.name AS product, p.category, s.revenue
FROM sales s
JOIN stores   st ON s.store_id   = st.store_id
JOIN products p  ON s.product_id = p.product_id
WHERE p.category = 'Tools'
ORDER BY s.revenue DESC
````

````python
%sql
-- ANSWER Q4 — WOS Core Metric
SELECT st.city, p.name AS product,
       i.stock_units, i.weekly_sales,
       ROUND(i.stock_units / i.weekly_sales, 1) AS wos,
       CASE
           WHEN i.stock_units / i.weekly_sales > 15 THEN 'OVERSTOCKED'
           WHEN i.stock_units / i.weekly_sales < 5  THEN 'LOW STOCK'
           ELSE 'HEALTHY'
       END AS status
FROM inventory i
JOIN stores   st ON i.store_id   = st.store_id
JOIN products p  ON i.product_id = p.product_id
ORDER BY wos DESC
````

````python
%sql
-- ANSWER Q6 — Forecast vs Actual MAPE
SELECT st.city, p.name AS product,
       f.forecast_month, f.forecasted_units, f.actual_units,
       f.actual_units - f.forecasted_units AS diff,
       ROUND(ABS(f.actual_units - f.forecasted_units) * 100.0 / f.actual_units, 1) AS mape,
       CASE WHEN ABS(f.actual_units - f.forecasted_units) * 100.0 / f.actual_units > 20
            THEN 'HIGH ERROR' ELSE 'OK' END AS flag
FROM forecast f
JOIN stores   st ON f.store_id   = st.store_id
JOIN products p  ON f.product_id = p.product_id
WHERE f.actual_units IS NOT NULL
ORDER BY mape DESC
````

````python
%sql
-- ANSWER Q8 — Conditional Aggregation
SELECT st.region,
       ROUND(SUM(s.revenue), 0) AS total_rev,
       ROUND(SUM(CASE WHEN st.size = 'Large'  THEN s.revenue ELSE 0 END), 0) AS large_rev,
       ROUND(SUM(CASE WHEN st.size = 'Medium' THEN s.revenue ELSE 0 END), 0) AS medium_rev,
       ROUND(SUM(CASE WHEN st.size = 'Small'  THEN s.revenue ELSE 0 END), 0) AS small_rev
FROM sales s
JOIN stores st ON s.store_id = st.store_id
GROUP BY st.region
ORDER BY total_rev DESC
````

````python
%sql
-- ANSWER Q18 — Basic CTE
WITH store_rev AS (
    SELECT s.store_id, st.city, SUM(s.revenue) AS total
    FROM sales s
    JOIN stores st ON s.store_id = st.store_id
    GROUP BY s.store_id, st.city
)
SELECT city, ROUND(total, 0) AS total_revenue
FROM store_rev
WHERE total > (SELECT AVG(total) FROM store_rev)
ORDER BY total DESC
````

````python
%sql
-- ANSWER Q24 — Top 2 Stores per Product
WITH agg AS (
    SELECT s.product_id, s.store_id, st.city,
           SUM(s.revenue) AS revenue
    FROM sales s
    JOIN stores st ON s.store_id = st.store_id
    GROUP BY s.product_id, s.store_id, st.city
),
ranked AS (
    SELECT *,
           ROW_NUMBER() OVER (PARTITION BY product_id ORDER BY revenue DESC) AS rn
    FROM agg
)
SELECT p.name AS product, r.city, r.revenue, r.rn AS rank
FROM ranked r
JOIN products p ON r.product_id = p.product_id
WHERE r.rn <= 2
ORDER BY p.name, r.rn
````

````python
%sql
-- ANSWER Q25 — LAG Velocity Shift
SELECT st.city, p.name AS product,
       s.sale_date, s.revenue,
       LAG(s.revenue) OVER (PARTITION BY s.store_id, s.product_id ORDER BY s.sale_date) AS prev_revenue,
       s.revenue - LAG(s.revenue) OVER (PARTITION BY s.store_id, s.product_id ORDER BY s.sale_date) AS change
FROM sales s
JOIN stores   st ON s.store_id   = st.store_id
JOIN products p  ON s.product_id = p.product_id
ORDER BY st.city, p.name, s.sale_date
````

````python
%sql
-- ANSWER Q30 — Most Overstocked City per Product
WITH wos_calc AS (
    SELECT i.product_id, st.city,
           ROUND(i.stock_units / i.weekly_sales, 1) AS wos
    FROM inventory i
    JOIN stores st ON i.store_id = st.store_id
),
ranked AS (
    SELECT *,
           RANK() OVER (PARTITION BY product_id ORDER BY wos DESC) AS rk
    FROM wos_calc
)
SELECT p.name AS product, r.city, r.wos AS highest_wos
FROM ranked r
JOIN products p ON r.product_id = p.product_id
WHERE r.rk = 1
ORDER BY r.wos DESC
````
